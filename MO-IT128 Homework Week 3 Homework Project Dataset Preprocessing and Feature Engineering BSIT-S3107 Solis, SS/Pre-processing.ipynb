```python
import numpy as np
import pandas as pd
```

```python
transactions = pd.read_csv("Transaction_Data.csv")
feedback = pd.read_csv("Customer_Feedback_Data.csv")
products = pd.read_csv("Product_Offering_Data.csv")
```

```python
transactions = transactions.drop_duplicates()
feedback = feedback.drop_duplicates()
products = products.drop_duplicates()
```

```python
transactions["Transaction_Date"] = pd.to_datetime(transactions["Transaction_Date"], errors="coerce")
```

```python
transactions["Transaction_Amount"] = pd.to_numeric(transactions["Transaction_Amount"], errors="coerce")
feedback["Satisfaction_Score"] = pd.to_numeric(feedback["Satisfaction_Score"], errors="coerce")
feedback["Likelihood_to_Recommend"] = pd.to_numeric(feedback["Likelihood_to_Recommend"], errors="coerce")
```

```python
transactions["Transaction_Type"] = transactions["Transaction_Type"].astype(str).str.strip().str.title()

products["Product_Name"] = products["Product_Name"].astype(str).str.strip()
products["Product_Type"] = products["Product_Type"].astype(str).str.strip().str.title()
products["Risk_Level"] = products["Risk_Level"].astype(str).str.strip().str.title()
products["Target_Age_Group"] = products["Target_Age_Group"].astype(str).str.strip()
products["Target_Income_Group"] = products["Target_Income_Group"].astype(str).str.strip()

feedback["Feedback_Comments"] = feedback["Feedback_Comments"].astype(str).str.strip()
```

```python
transactions["Transaction_Date"] = transactions["Transaction_Date"].fillna(transactions["Transaction_Date"].mode()[0])
transactions["Transaction_Amount"] = transactions["Transaction_Amount"].fillna(transactions["Transaction_Amount"].median())
transactions["Transaction_Type"] = transactions["Transaction_Type"].replace(["Nan", "None", ""], np.nan)
transactions["Transaction_Type"] = transactions["Transaction_Type"].fillna(transactions["Transaction_Type"].mode()[0])
```

```python
feedback["Satisfaction_Score"] = feedback["Satisfaction_Score"].fillna(feedback["Satisfaction_Score"].median())
feedback["Likelihood_to_Recommend"] = feedback["Likelihood_to_Recommend"].fillna(feedback["Likelihood_to_Recommend"].median())
feedback["Feedback_Comments"] = feedback["Feedback_Comments"].replace(["nan", "None", ""], np.nan)
feedback["Feedback_Comments"] = feedback["Feedback_Comments"].fillna("No comment")
```

```python
for col in ["Product_Name", "Product_Type", "Risk_Level", "Target_Age_Group", "Target_Income_Group"]:
    products[col] = products[col].replace(["nan", "None", ""], np.nan)
    products[col] = products[col].fillna("Unknown")
```

```python
merged_data = transactions.merge(feedback, on="Customer_ID", how="left")
```

```python
merged_data["Satisfaction_Score"] = merged_data["Satisfaction_Score"].fillna(merged_data["Satisfaction_Score"].median())
merged_data["Likelihood_to_Recommend"] = merged_data["Likelihood_to_Recommend"].fillna(merged_data["Likelihood_to_Recommend"].median())
merged_data["Feedback_Comments"] = merged_data["Feedback_Comments"].fillna("No feedback")
```

```python
transactions.to_csv("transactions_cleaned.csv", index=False)
feedback.to_csv("feedback_cleaned.csv", index=False)
products.to_csv("products_cleaned.csv", index=False)
merged_data.to_csv("merged_transactions_feedback.csv", index=False)
```
