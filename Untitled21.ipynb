```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("merged_transactions_feedback.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
upper_limit = df["Transaction_Amount"].quantile(0.95)

plt.figure(figsize=(8,5))
plt.hist(df[df["Transaction_Amount"] <= upper_limit]["Transaction_Amount"], bins=50)
plt.title("Distribution of Transaction Amount (Up to 95th Percentile)")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
upper_sat = df["Satisfaction_Score"].quantile(0.95)

plt.figure(figsize=(8,5))
plt.hist(df[df["Satisfaction_Score"] <= upper_sat]["Satisfaction_Score"], bins=20)
plt.title("Distribution of Satisfaction Score (Up to 95th Percentile)")
plt.xlabel("Satisfaction Score")
plt.ylabel("Frequency")
plt.show()
```

```python
transaction_counts = df["Transaction_Type"].value_counts().sort_values()

plt.figure(figsize=(8,5))
transaction_counts.plot(kind="bar")
plt.title("Transaction Type Distribution (Zoomed View)")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")

# Zoom the y-axis
plt.ylim(transaction_counts.min() - 200, transaction_counts.max() + 200)

plt.xticks(rotation=30)
plt.show()
```

```python
upper_amt = df["Transaction_Amount"].quantile(0.95)

plt.figure(figsize=(8,5))
plt.scatter(
    df[df["Transaction_Amount"] <= upper_amt]["Transaction_Amount"],
    df[df["Transaction_Amount"] <= upper_amt]["Satisfaction_Score"],
    alpha=0.5
)
plt.title("Transaction Amount vs Satisfaction Score (Up to 95th Percentile)")
plt.xlabel("Transaction Amount")
plt.ylabel("Satisfaction Score")
plt.show()
```
